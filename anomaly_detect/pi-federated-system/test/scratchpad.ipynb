```python
dummy_input = np.random.normal(0, 1, (1, 160))
initial_score = model.predict(dummy_input)
print(f"Prediction score (initial): {initial_score: .4f}")
```

```python
training_data = np.random.normal(0, 0.1, (100, 160))
loss = model.train(training_data)
print(f"Training loss: {loss: .4f}")
```

```python
normal_score = model.predict(np.random.normal(0, 0.1, (1, 160)))
anomaly_score = model.predict(np.random.normal(5, 10, (1, 160)))

print(f"Post-train normal score: {normal_score: .4f}")
print(f"Post-train anomaly score: {anomaly_score: .4f}")
```
